## Исправление
Что произошло:  мы попытались использовать переменную project_title, но программа не нашла ее в памяти (runtime), потому чтго мы ее еще не создали. Суть исправленпия: сначала мы должны создать переменную, а потом ее использовать.

```python
project_title = "Text Lab"
```

Вывод по базовому уровню: Порядок выполнения ячеек имеет решающее значение. Переменные существуют только в текущей сессии (runtime). При перезапуске ядра все переменные теряютя - это нужно учитывать при работе.
